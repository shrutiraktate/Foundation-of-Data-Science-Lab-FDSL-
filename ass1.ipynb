```python
import numpy as np
print("NumPy version:", np.__version__)
```

```python
# 1D array from a list
arr1 = np.array([10, 52, 38, 74, 95])
print("1D array:", arr1)
```

```python
# 2D array from a nested list (matrix)
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array:\n", arr2)
```

```python
# 3D array
arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D array:\n", arr3)
```

```python
# Array of zeros
zeros_arr = np.zeros((2, 3))
print("Zeros array:\n", zeros_arr)
```

```python
# Array of ones
ones_arr = np.ones((3, 2))
print("Ones array:\n", ones_arr)
```

```python
# Array filled with a constant value
full_arr = np.full((2, 2), 7)
print("Full array:\n", full_arr)
```

```python
# arange -> similar to Python range() but returns an array
range_arr = np.arange(0, 10, 2)   # start, stop, step
print("arange array:", range_arr)

```

```python
# Random arrays
rand_arr = np.random.rand(2, 3)      # uniform distribution [0, 1)
randint_arr = np.random.randint(1, 100, size=(2, 3))  # random integers
print("Random float array:\n", rand_arr)
print("Random integer array:\n", randint_arr)
```

```python
exercise1_a = np.zeros((4, 4))
exercise1_b = np.arange(10, 51, 5)
print(exercise1_a)
print(exercise1_b)
```

```python
sample = np.array([[1, 2, 3], [4, 5, 6]])

print("Array:\n", sample)
print("Shape:", sample.shape)     
print("Dimensions:", sample.ndim)  
print("Size:", sample.size)        
print("Data type:", sample.dtype)  
```

```python
# Sorting arrays
unsorted = np.array([5, 2, 9, 1, 7])
sorted_arr = np.sort(unsorted)
print("Original:", unsorted)
print("Sorted:", sorted_arr)
```

```python
ex2 = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape:", ex2.shape)
print("Size:", ex2.size)
print("Reshaped:\n", ex2.reshape(3, 2))
```

```python
x = np.array([10, 20, 30, 40])
y = np.array([1, 2, 3, 4])

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
print("Power:", x ** 2)
```

```python
data = np.array([4, 8, 15, 16, 23, 42])

print("Sum:", np.sum(data))
print("Mean:", np.mean(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Standard deviation:", np.std(data))
print("Variance:", np.var(data))
```

```python
# Universal functions work element-wise
nums = np.array([1, 4, 9, 16])

print("Square root:", np.sqrt(nums))
print("Exponential:", np.exp(nums))
print("Natural log:", np.log(nums))
print("Sine:", np.sin(nums))
```

```python
# Matrix operations
m1 = np.array([[1, 2], [3, 4]])
m2 = np.array([[5, 6], [7, 8]])

print("Element-wise multiplication:\n", m1 * m2)
print("Matrix multiplication (dot):\n", np.dot(m1, m2))
print("Matrix multiplication (matmul):\n", np.matmul(m1, m2))
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print("Full array:", arr)
print("First element:", arr[0])
print("Last element:", arr[-1])
print("Elements from index 1 to 3:", arr[1:4])
print("Every second element:", arr[::2])
print("Reversed array:", arr[::-1])
```
