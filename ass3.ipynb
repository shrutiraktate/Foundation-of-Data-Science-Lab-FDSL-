```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("StudentsPerformance.csv")
```

```python
print(df.head())
```

```python
print(df.describe())
```

```python
mean_val = df['math score'].mean()
median_val = df['math score'].median()
mode_val = df['math score'].mode()[0]
```

```python
print('Mean:', mean_val)
print('Median:', median_val)
print('Mode:', mode_val)
```

```python
marks_array = df['math score'].to_numpy()
```

```python
print('NumPy Mean:', np.mean(marks_array))
print('NumPy Median:', np.median(marks_array))
print('SciPy Mode:', stats.mode(marks_array, keepdims=True))
```

```python
group_summary = df.groupby('gender')['math score'].agg([
    'mean',
    'median',
    lambda x: x.mode()[0]
])
```

```python
group_summary.columns = ['Mean', 'Median', 'Mode']
```

```python
print("\nGroup-wise Analysis:")
print(group_summary)
```

```python
print('\nSkewness:', df['math score'].skew())

if mean_val > median_val:
    print('Distribution is right (positively) skewed')
elif mean_val < ,median_val:
    print('Distribution is left (negatively) skewed')
else:
    print('Distribution is approximately symmetric')
```
